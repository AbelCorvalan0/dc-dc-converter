## 2. Half-Bridge


### 2.1 Input current $I_{in}$ (per module)

Assuming:

$$ \eta= 95\% $$

Input Power:

$$ \boxed{ P_{in}= \frac{P_{per \, module}}{\eta} }$$

Average current:

$$\boxed{ I_{in} = \frac{P_{in}}{V_{in}} }$$

```python
eta = 95 #%
# Input power
Pin = Pmod/(eta/100)
# Input current
Iin = Pin/Vin

print("Corriente de entrada Iin= ", Iin, "A")
```

### 2.2 Input inductance $L_{in}$

$\frac{di}{dt}$ limitation design:

$$ L = \frac{V}{di/dt} $$

<!-- We determinate:

$$ \frac{di}{dt} = 1000 \frac{A}{s} $$ -->


Alternative design:

$$ L_{in} = \frac{V_{in}D}{\Delta I_{L} f_{s}} $$

```python
Il_perc = 0.2
# 2.5
# 0.02
# Approximated percentage
dIl = Iin*(Il_perc/100)
#dIl

Lin = (Vin*D)/(fs*dIl)
# 45mHy
print("Valor de inductancia de entrada, Lin=", Lin*1e6, "uHy")

# Peak input inductance current
#Ipk = Iin + (dIl/2)
#print("Corriente pico en inductor de entrada Ipeak_in=", Ipk, "A \n")
#print("El núcleo debe soportar ese valor de Ipeak_in sin saturar")
```

### 2.3 Half-Brdige capacitors $C_{U}$ y $C_{L}$

We have:

$$ V_{bus} = V_{C_{U}} + V_{C_{L}} $$

$$ V_{bus} \approx 2V_{in}$$

Assuming:

$$ \Delta V_{bus} = 2V $$


$$ C_{U} = C_{L} = \frac{I_{in}}{2 \, f_{s} \Delta V} $$

### 2.1 DC bus voltage


```python
# Vprim está afectada por Llkp y Lmp

Vbus = 2*Vin

# 2V
dVbus = 2 # Volt
dVin  = 1 # Volt

Cu = Iin/(2*fs*dVin)
Cl = Cu
#Cu= 346e-6 (estaba en simulación)
# 135e-6
print("El valor de la tensión del bus Vbus es: ", Vbus, "V")
print("El valor de los capacitores de entrada (Cu, Cl) es: ", Cl*1e6, "uF")
```

### 2.2 Resonant capacitor $C_{r}$ (parallel with MOSFET)

$$ f_{r} = \frac{ 1 }{ 2\pi \sqrt{L_{lkp}C_{r}} } $$


$$ C_{r} = \frac{1}{L_{lkp} (2\pi f_{r})^{2}} $$

```python
#Lmp = 0.00001 # 10uHy

# Delta magnetization current
dIm = 1

Lmp = (Vbus*D)/(2*fs*dIm)

#Lmp = np.linspace(0.1e-6, 100e-6, 100)
#Llkp = 0.0003

# Percentage (2% to 10%)
Llkp_perc = 90
#Llkp_perc = np.linspace(2, 10, 100)

Llkp = (Llkp_perc/100)*Lmp

print("Llkp =", Llkp*1e6, "uHy\nLmp =", Lmp*1e6,"uHy")
# EN SIMULACIÓN
# Llkp = 2e-6
# Lm   = 1e-3
# En inductor decía 100e-6

# Resonance capacitor (parallel with MOSFET)
Cr = 1/(Llkp*(2*np.pi*fs)**2)
print("Valor de Capacitor de MOSFET: ", Cr, "F")
```

### Verification of ZVS (Zero-Voltage Switching)

The energy stored in $L_{lkp}$ and $L_{mp}$ must be sufficient to charge/discharge MOSFET capacitors.
Condition:

$$ \frac{1}{2} L_{eq} I^{2}_{p} > \frac{1}{2} C_{oss} V^{2}_{bus} $$

- $L_{eq}$ is the resonant inductance involved.
- $C_{oss}$ is MOSFET output capacitance.

Where $L_{eq}$ is calculated as:

$$ L_{eq} = L_{lkp} + L_{mp} $$

$$ L_{eq} \approx L_{lkp} $$

```python
#Leq = Llkp + Lmp
Leq = Llkp + Lmp
print(Leq)

E_inductive  = (Leq/2)*(Ipk)**(2)

E_capacitive = (Cr/2)*(Vbus)**2

print("Energía almacenada en inductor (Llkp): ", E_inductive*1e3, "mJ")
print("Energía almacenada en capacitores (Cr): ", E_capacitive*1e3, "mJ\n\n")

if (E_inductive > E_capacitive):
    print("ZVS (Zero-Voltage Switching) CORRECTO")
else:
    print("NO SE VERIFICA ZVS (Zero-Voltage Switching)")
```
